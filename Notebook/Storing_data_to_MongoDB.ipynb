
### Assumption and Data Handling Process

I am assuming that the CSV file is not available and instead, the data is stored in MongoDB. The steps I followed are:

1. **Dataset Structure**: The data is stored in MongoDB under the **Shipment** database, with a collection named **Shipping data**.

2. **Converting DataFrame**: I used Python to convert the Pandas DataFrame into a dictionary format using `to_dict()` to make it compatible with MongoDB.

3. **MongoDB Insertion**: The converted data is inserted into the **Shipping data** collection using the `insert_many()` method, mapping DataFrame columns to MongoDB document fields.

4. **Python Notebook**: The process is handled in a notebook, **Storing Data to MongoDB.ipynb**, where the data is read, converted, and inserted into the collection.
